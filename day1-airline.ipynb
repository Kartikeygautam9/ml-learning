```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import LabelEncoder

train = pd.read_csv('/kaggle/input/competitions/playground-series-s6e10/train.csv')
test  = pd.read_csv('/kaggle/input/competitions/playground-series-s6e10/test.csv')

# Target ko 0/1 numbers mein badlo
le = LabelEncoder()
y = le.fit_transform(train['satisfaction'])
print("Classes:", list(le.classes_))

# Features banao
X  = pd.get_dummies(train.drop(['id', 'satisfaction'], axis=1))
Xt = pd.get_dummies(test.drop('id', axis=1)).reindex(columns=X.columns, fill_value=0)

# Missing values ko 0 se bhar do
X  = X.fillna(0)
Xt = Xt.fillna(0)

# Train + predict
model = LogisticRegression(max_iter=1000)
model.fit(X, y)
probs = model.predict_proba(Xt)[:, 1]

# Submission
sub = pd.DataFrame({'id': test['id'], 'satisfaction': probs})
sub.to_csv('submission.csv', index=False)
print(sub.head())
```
